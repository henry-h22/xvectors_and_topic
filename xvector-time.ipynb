```python
from model_loaders import *
classifier = get_classifier()
wav_dir = 'data/wav' # these will change w.r.t. the BNHP database
eaf_dir = 'data/eaf'
output_dir = 'output'
```

```python
for file in Path(wav_dir).iterdir():
    if file.is_file() and file.name.endswith('.wav'):
        title = file.name.split('.')[0]
        embeddings = title_to_embeddings(title, classifier, wav_dir, eaf_dir)
        save_embedded_batch(embeddings, title, output_dir)
```
